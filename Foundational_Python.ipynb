# Stage 1.3 Cleaning

```python
# Using dropna() to remove any rows with null values
transfers_no_nulls = transfers.dropna()
transfers_no_nulls.info()
```

```python
# Using fillna() to fill in any null values with a value
transfers['transfer_fee'] = transfers['transfer_fee'].fillna(0)
transfers['market_value_in_eur'] = transfers['market_value_in_eur'].fillna(0)
transfers.info()
```

```python
# Using astype() to change data type of a column
transfers['player_id'] = transfers['player_id'].astype('str')
transfers.info()
```

```python
# Using drop_duplicates() to remove any duplicates and get unqiue values
buying_club = transfers['to_club_name']
buying_club_unique = buying_club.drop_duplicates().reset_index()
buying_club_unique
```

```python
#Dropping duplicated rows
transfers = transfers.drop_duplicates()
transfers.info()
```

```python
#Getting latest transfer for each player
transfers = transfers.sort_values('transfer_date', ascending = False)
transfers_unique_players = transfers.drop_duplicates(subset = 'player_id')
transfers_unique_players.info()
```

```python
#Using rename() to rename columns
transfers_updated = transfers.rename({'from_club_name': 'selling_club','to_club_name': 'buying_clubs'}, axis=1)
transfers_updated.info()
```

```python
transfers_updated['Transfer_type'] = "Free Agent"
transfers_updated.loc[(transfers_updated['transfer_fee'] == 0) & (transfers_updated['market_value_in_eur'] == 0),'Transfer_type'] = 'Free Agent'
transfers_updated.loc[(transfers_updated['transfer_fee'] == 0) & (transfers_updated['market_value_in_eur'] != 0),'Transfer_type'] = 'Free Transfer'
transfers_updated.loc[(transfers_updated['transfer_fee'] != 0), 'Transfer_type'] = 'Paid Transfer'
transfers_updated['Transfer_type'] = pd.Categorical(transfers_updated['Transfer_type'], ['Free Agent','Free Transfer','Paid Transfer'])
transfers_updated.dtypes
```

# Stage 1.4 Row Navigation

```python
# Using .shift() to get the previous year/month/day value

transfers_for_clubs =  transfers_updated[['transfer_date','buying_clubs','transfer_fee']]
transfers_for_clubs['transfer_date'] = pd.to_datetime(transfers_for_clubs['transfer_date'])
transfers_for_clubs['Year'] = transfers_for_clubs['transfer_date'].dt.year
transfers_manc = transfers_for_clubs[transfers_for_clubs['buying_clubs'] == 'Man City']
transfers_manc = transfers_manc.groupby('Year', as_index = False).agg({'transfer_fee':'sum'}).sort_values('Year',ascending = True)
transfers_manc['Previouzs_year_transfer_fee'] = transfers_manc['transfer_fee'].shift(1)
transfers_manc.head()
```

```python
# Using .diff() to get the yearly/monthly/daily difference
transfers_manc['YoY_difference'] = transfers_manc['transfer_fee'].diff()
transfers_manc.head()
```
