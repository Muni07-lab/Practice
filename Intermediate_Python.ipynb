# Stage 2.2 Group By & Aggregation

```python
def result(x,y):
    if x > y:
        return 'Home Win'
    elif x == y:
        return 'Draw'
    elif x < y:
        return 'Away Win'
    else:
        return None
```

```python
def win_draw_loss(x,y):
    if x == 'Draw':
        return 'D'
    elif (x == 'Home Win' and y == 'Home'):
        return 'W'
    elif (x == 'Away Win' and y == 'Away'):
        return 'W'
    elif (x == 'Home Win' and y == 'Away'):
        return 'L'
    elif (x == 'Away Win' and y == 'Home'):
        return 'L'
    else:
        return None
```

```python
def point(x):
    if x == 'W':
        return 3
    elif x == 'D':
        return 1
    elif x == 'L':
        return 0
    else:
        return None
```

```python
games['Result'] = games[['home_club_goals','away_club_goals']].apply(
    lambda row: result(row['home_club_goals'], row['away_club_goals']),
    axis=1
)

home_games = games[['game_id','date','competition_id','season','round','home_club_id','home_club_goals','stadium','attendance','referee','home_club_name','competition_type','Result']].rename({'home_club_id':'club_id','home_club_goals':'goals','home_club_name':'club_name'},axis=1)
away_games = games[['game_id','date','competition_id','season','round','away_club_id','away_club_goals','stadium','attendance','referee','away_club_name','competition_type','Result']].rename({'away_club_id':'club_id','away_club_goals':'goals','away_club_name':'club_name'},axis=1)
home_games['Location'] = 'Home'
away_games['Location'] = 'Away'
all_games = pd.concat([home_games, away_games])
all_games['result'] = all_games[['Result','Location']].apply(
    lambda row: win_draw_loss(row['Result'], row['Location']),
    axis=1
)
all_games
```

```python
# Using GroupBy and agg
Club_results = all_games.groupby('club_name').agg({'game_id':'count','goals':'mean'}).reset_index()
Club_results
#Club_results[Club_results['club_name'] == 'England']
```

```python
# Using apply for rows
all_games['Points'] = all_games['result'].apply(point)
all_games.head(5)
```

```python
file_path_sales = "/Users/munavwarhussain/Library/Mobile Documents/com~apple~CloudDocs/Career Work/Job Preperation/Technical Tools & Roles/Python/Python Data Analyst Skill Check/Data/sales.csv"
sales = pd.read_csv(file_path_sales)
sales.info()
```

```python
# Creating date range
dates = pd.date_range(start = sales['date'].min(), end = sales['date'].max(), freq = 'D')
dates
```

```python
# Changing time series data to change frequency 
sales['date'] = pd.to_datetime(sales['date'])
sales_resampled_month = sales.resample('M', on = 'date').sum()[['sales_qty','sales_revenue']].reset_index()
print(sales_resampled_month.head())
sales_resampled_year = sales.resample('Y', on = 'date').sum()[['sales_qty','sales_revenue']].reset_index()
print(sales_resampled_year.head())
```

```python
# Making a date column an index
sales = sales.set_index(sales['date'])
sales.head()
```

```python
# Find Previous months value, MoM difference and percentage change
sales_resampled_month['Year-Month'] = sales_resampled_month['date'].dt.strftime('%Y-%m')
sales_resampled_month['Previous_Month_Revenue'] = sales_resampled_month['sales_revenue'].shift(1)
sales_resampled_month['MoM_difference'] = sales_resampled_month['sales_revenue'].diff(1)
sales_resampled_month['MoM_pct_change'] = sales_resampled_month['sales_revenue'].pct_change(1)*100
sales_resampled_month.head()
```

```python
# Upsampling to get sales for every day
sales_resampled_month = sales_resampled_month[['sales_qty','sales_revenue','Year-Month']]
sales_resampled_month['Year-Month'] = pd.to_datetime(sales_resampled_month['Year-Month'])
sales_resampled_month = sales_resampled_month.set_index('Year-Month')
sales_resampled_day = sales_resampled_month.asfreq('D', method=None)
sales_resampled_day
```

```python
# Front fill
sales_resampled_day_ffill = sales_resampled_day.ffill()
sales_resampled_day_ffill.head(13)
```

```python
#Interpolate
sales_resampled_day_interpolate = sales_resampled_day.interpolate()
sales_resampled_day_interpolate.head(13)
```

```python
# Cumulative sum and Cumulative product
sales_resampled_month['Cumulative sales'] = sales_resampled_month['sales_revenue'].cumsum()
sales_resampled_month['perc_change'] = sales_resampled_month['sales_revenue'].pct_change(1)
sales_resampled_month['perc_change'] = sales_resampled_month['perc_change'] + 1
sales_resampled_month['Growth_factor'] = sales_resampled_month['perc_change'].cumprod()
sales_resampled_month.head()
```

# Stage 2.9 Regex mastery

```python
# Use re.compile to create regex match object
import re 
regex_match = re.compile(r'[\d]{4}-[\d]{2}-[\d]{2}')
```

```python
# Use re.search() to create a match object and either produce the match or test the match
transfers['Date_check'] = transfers['transfer_date'].apply(lambda x: bool(re.search(regex_match,x)))
transfers['Actual_date'] = transfers['transfer_date'].apply(lambda x: re.search(regex_match,x).group())
print(transfers['Date_check'].value_counts())
transfers['Actual_date'].head(2)
```

```python
# Use re.match() to create a match object with the start of the string
transfers['Year'] = transfers['transfer_date'].apply(lambda x: re.search(r'\d{4}',x).group())
transfers['Year'].head(2)
```

```python
# Use re.fullmatch() to create a match object that must match the exact string
transfers['Date_check'] = transfers['transfer_date'].apply(lambda x: re.fullmatch(r'\d{4}-\d{2}-\d{2}',x).group())
transfers['Date_check'].head(2)
```

```python
# Use re.findall() to create a list of the multple matches of the string
split_date = transfers['transfer_date'].apply(lambda x: re.findall(r'[\d]+',x))
transfers['Year'] = split_date.apply(lambda x: x[0])
transfers['Month'] = split_date.apply(lambda x: x[1])
transfers['Day'] = split_date.apply(lambda x: x[2])

print(transfers['Year'].head(2))
print(transfers['Month'].head(2))
print(transfers['Day'].head(2))
```

```python
# Use re.findall() to create a list of the multple matches of the string
split_date = transfers['transfer_date'].apply(lambda x: re.split(r'-',x))
transfers['Year'] = split_date.apply(lambda x: x[0])
transfers['Month'] = split_date.apply(lambda x: x[1])
transfers['Day'] = split_date.apply(lambda x: x[2])

print(transfers['Year'].head(2))
print(transfers['Month'].head(2))
print(transfers['Day'].head(2))
```

```python
# Use re.sub() to clean string column
transfers['transfer_date'] = transfers['transfer_date'].apply(lambda x: re.sub(r'/',"-",x))
transfers.head(2)
```
