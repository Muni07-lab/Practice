# Stage 2.2 Group By & Aggregation

```python
def result(x,y):
    if x > y:
        return 'Home Win'
    elif x == y:
        return 'Draw'
    elif x < y:
        return 'Away Win'
    else:
        return None
```

```python
def win_draw_loss(x,y):
    if x == 'Draw':
        return 'D'
    elif (x == 'Home Win' and y == 'Home'):
        return 'W'
    elif (x == 'Away Win' and y == 'Away'):
        return 'W'
    elif (x == 'Home Win' and y == 'Away'):
        return 'L'
    elif (x == 'Away Win' and y == 'Home'):
        return 'L'
    else:
        return None
```

```python
def point(x):
    if x == 'W':
        return 3
    elif x == 'D':
        return 1
    elif x == 'L':
        return 0
    else:
        return None
```

```python
games['Result'] = games[['home_club_goals','away_club_goals']].apply(
    lambda row: result(row['home_club_goals'], row['away_club_goals']),
    axis=1
)

home_games = games[['game_id','competition_id','season','round','home_club_id','home_club_goals','stadium','attendance','referee','home_club_name','competition_type','Result']].rename({'home_club_id':'club_id','home_club_goals':'goals','home_club_name':'club_name'},axis=1)
away_games = games[['game_id','competition_id','season','round','away_club_id','away_club_goals','stadium','attendance','referee','away_club_name','competition_type','Result']].rename({'away_club_id':'club_id','away_club_goals':'goals','away_club_name':'club_name'},axis=1)
home_games['Location'] = 'Home'
away_games['Location'] = 'Away'
all_games = pd.concat([home_games, away_games])
all_games['result'] = all_games[['Result','Location']].apply(
    lambda row: win_draw_loss(row['Result'], row['Location']),
    axis=1
)
all_games
```

```python
# Using GroupBy and agg
Club_results = all_games.groupby('club_name').agg({'game_id':'count','goals':'mean'}).reset_index()
Club_results
#Club_results[Club_results['club_name'] == 'England']
```

```python
# Using apply for rows
all_games['Points'] = all_games['result'].apply(point)
all_games.head(5)
```

# Stage 2.3 Merging & joining

```python
# Merge example
player_games = appearances.merge(all_games, left_on = ['game_id','player_club_id'],right_on = ['game_id','club_id'], how = 'left')
player_games.sort_values('game_id').head()
```
